```python
def segment_customer(score):
    if score >= 13:
        return "VIP"
    elif score >= 10:
        return "Loyal"
    elif score >= 7:
        return "Potential Loyal"
    elif score >= 5:
        return "At Risk"
    else:
        return "Lost"
```

```python
customer_data["Segment"] = customer_data["RFM_Score"].apply(
    segment_customer
)
```
